```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

text = ['I would not like them here or there.', 'I would not like them anywhere.', 'I do not like green eggs and ham.', 'I do not like them, Sam-I-am.']

vectorizer = CountVectorizer(token_pattern=r"(?u)\b\w[\w-]*\w\b|\b\w\b")
vectors = vectorizer.fit_transform(text)

df = pd.DataFrame(data=vectors.toarray(), columns=vectorizer.get_feature_names_out(), index=[f"Sentence {i+1}" for i in range(len(text))])
print(df)
```

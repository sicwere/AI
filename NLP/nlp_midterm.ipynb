```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
import numpy as np

sentence = "You want to wake up in the morning and think the future is going to be great - and that’s what being a spacefaring civilization is all about. It’s about believing in the future and thinking that the future will be better than the past. And I can’t think of anything more exciting than going out there and being among the stars."
tokens = sorted(set(sentence.lower().split()))
data = np.array(list(tokens)).reshape(-1, 1)
encoder = OneHotEncoder(
    sparse_output=False,       
    handle_unknown='ignore' 
)

encoder.set_output(transform='pandas')
df = encoder.fit_transform(data)
print(df)
```

```python
import requests
from bs4 import BeautifulSoup
import re

url = "https://www.nationalacademies.org/read/26809/chapter/1"
headers = {"User-Agent": "Mozilla/5.0"}  # Mimics a real browser to prevent blocking
response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

for tag in soup(["script", "style", "nav", "footer", "header", "aside"]):
    tag.decompose()

clean_text = re.split(r'Page', soup.get_text(separator="\n", strip=True))

print(clean_text)
```

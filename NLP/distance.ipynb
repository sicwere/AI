```python
from gensim.models import Word2Vec
import numpy as np

model = Word2Vec(sentences=[['beautiful', 'beauty']], vector_size=100, window=5, min_count=1, workers=4)

beauty = model.wv['beauty']
beautiful = model.wv['beautiful']

cosine_similarity = model.wv.similarity('beautiful', 'beauty')

print(f"Euclidean Distance: {np.linalg.norm(beautiful - beauty)}")
print(f'Cosine Similarity: {cosine_similarity}')
print(f'Cosine Difference: {1 - cosine_similarity}')
```

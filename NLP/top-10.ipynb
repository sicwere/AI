```python
import glob
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
import random
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, ConfusionMatrixDisplay

files = glob.glob("./top-10/*.csv", recursive=True)
df = pd.concat((pd.read_csv(f) for f in files), ignore_index=True)

df['review'] = df['review'].str.lower()
reviews = list(df[['review', "rating"]].dropna().to_records(index=False))
random.shuffle(reviews)
train_documents, test_documents = train_test_split(reviews, test_size=0.2, random_state=42)
X_train, y_train = zip(*train_documents)
X_test, y_test = zip(*test_documents)

vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

clf = MultinomialNB()
clf.fit(X_train_tfidf, y_train)

y_pred = clf.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

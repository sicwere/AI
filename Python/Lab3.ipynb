### Task #1: Create a list of customers using data from file Customers.txt (using the list data structure) (3 points)

Write code that will read the text file Customers.txt and write the values from the file to a Python list. You must utilize the list data structure to complete this task.

Your code should print the following:

['Mary', 'Anna', 'Emma', 'Elizabeth', 'Margaret', 'Minnie', 'Clara', 'Bertha', 'Ida', 'Annie', 'Alice']

```python
# Write your code here - insert more code cells as needed
customers = []
with open('week3_datastructures_functions_files_codesamples\\week3_datastructures_functions_files\\Customers.txt', 'r') as file:
    for line in file:
        customers.append(line.strip())
print(customers)
```

### Task #2: Using the list above create two sets (using the set data structure) (2 points)

Write code that will take the values from the list created above and add them to two different sets. You must utilize the set data structure to complete this task.

#### The set criteria is as follows:
#### set1 - Includes customers with a name length smaller than or equal to 5 characters
#### set2 - Includes customers with a name length greater than or equal to 5 characters

Your code should print the following:

{'Clara', 'Alice', 'Emma', 'Mary', 'Anna', 'Ida', 'Annie'}  
{'Clara', 'Bertha', 'Margaret', 'Elizabeth', 'Alice', 'Minnie', 'Annie'}

Please note that the name order in each set may vary with your sets, but you should have the same names in your sets.


```python
# Write your code here - insert more code cells as needed
set1 = set()
set2 = set()
with open('week3_datastructures_functions_files_codesamples\\week3_datastructures_functions_files\\Customers.txt', 'r') as file:
    for line in file:
        line = line.strip()
        length = len(line)
        if length <= 5:
            set1.add(line)
        if length >= 5:
            set2.add(line)
print(set1)
print(set2)
```

### Task #3: Using set data structure methods, print customers that are common in set1 and set2. Also print customers that are unique in each set (2 points)

You must utilize set methods to complete this task.

Your code should print the following:

   - {'Alice', 'Clara', 'Annie'}
   - {'Ida', 'Anna', 'Mary', 'Emma'}
   - {'Bertha', 'Elizabeth', 'Minnie', 'Margaret'}
   
Please note that the name order in each set may vary with your sets, but you should have the same names in your sets.

```python
# Write your code here - insert more code cells as needed
set1 = set()
set2 = set()
with open('week3_datastructures_functions_files_codesamples\\week3_datastructures_functions_files\\Customers.txt', 'r') as file:
    for line in file:
        line = line.strip()
        length = len(line)
        if length <= 5:
            set1.add(line)
        if length >= 5:
            set2.add(line)
print(set1.intersection(set2))
print(set1.difference(set2))
print(set2.difference(set1))
```

### Task #4: Write customer names in set1 to Customersset1.txt and customer names in set2 to Customersset2.txt (3 points)

Write code that creates 2 new file called Customersset1.txt and Customersset2.txt. Write all of the customers from set1 to the Customersset1.txt file and write all of the customers from set2 to the Customersset2.txt file. The names in the files should match their respective sets.

```python
# Write your code here - insert more code cells as needed
set1 = set()
set2 = set()
with open('week3_datastructures_functions_files_codesamples\\week3_datastructures_functions_files\\Customers.txt', 'r') as file:
    for line in file:
        line = line.strip()
        length = len(line)
        if length <= 5:
            set1.add(line)
        if length >= 5:
            set2.add(line)
with open('Customersset1.txt', 'w') as file:
    for c in set1:
        file.write(f"{c}\n")
with open('Customersset2.txt', 'w') as file:
    for c in set2:
        file.write(f"{c}\n")
```

### Task #1: Apply 1D Numpy array concepts (2 points)

1. Create a 1D numpy array that contains int values (4, 3, 1, 33, 21, 67, 27, 89, 34, 67, 99, 89, 12, 43).  
``Hint: You can put these values in a python list and start from there.``


2. Write code that prints values (67, 27, 89, 34).  
You must use the colon : based slicing approach to complete this task.  
Your code should print the following:
   - [67 27 89 34]


3. Write code to print the average (mean) and median values for this array.  
You code should print the following:
   - 42.07142857142857 for the mean
   - 33.5 for the median

````python
# Write your code here - insert more code cells as needed

import numpy as np
nums = np.array([4, 3, 1, 33, 21, 67, 27, 89, 34, 67, 99, 89, 12, 43])
print(nums[5:9])
print(np.mean(nums))
print(np.median(nums))
````

### Task #2: Use 1D Numpy array to store string values (1 point)
1. Create a 1D numpy array that stores values (4, 3, 1, 33, 21, 67, 27, 89, 34, 67, 99, 89, 12, 43) AS STRING VALUES.  
`` Hint: Explore using dtype``  
Your code should print the following:
   - ['4' '3' '1' '33' '21' '67' '27' '89' '34' '67' '99' '89' '12' '43']

### Task #3: Apply 2D Numpy array concepts (2 points)

### Task #5: Combine 2D Numpy arrays (1 point)

You are given two numpy arrays in the following cell: x_2d & y_2d. Write code that uses the x_2d & y_2d arrays to produce the output shown below.  
Note that the new array shown below is stored in variable named "z_2d"

```
z_2d: 
 [[10 11 12 13 14 30 31 32 33]
 [15 16 17 18 19 34 35 36 37]
 [20 21 22 23 24 38 39 40 41]]
 ```

````python
import numpy as np
x_2d = np.arange(10, 25).reshape(3,5)
print("x_2d array: \n", x_2d)

y_2d = np.arange(30, 42).reshape(3,4)
print("\ny_2d array: \n", y_2d)
````

````python
# Write your code here - insert more code cells as needed
import numpy as np
z_2d = np.concatenate((x_2d, y_2d), axis=1)
print("\nz_2d array: \n", z_2d)
````

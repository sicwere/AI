### Task #1: Use if..else control flow for BMI calculations (1 point)
 Write code that calculates BMI based on user provided height and weight values in metric. 
 
 Your code should first ask the user to input their weight in kilograms. Then it should ask the user to input their height in meters. Both must utilize the input function.
 
 Not only should your code output the BMI numerical value, it should also output the BMI category which can be referenced below.
 
 To check your code, use inputs (not hardcoded values) of 85 kg for weight and 2 meters for height. Your code should print 21.25 and "Normal weight." 
 
 #### BMI Categories: 
Underweight: BMI < 18.5

Normal weight: – 18.5 <= BMI < 25

Overweight: 25 <= BMI < 30

Obesity: BMI >= 30

```python
# Write your code here - insert more code cells as needed
weight = float(input('Enter your weight in kilograms: '))
height = float(input('Enter your height in meters: '))
bmi = weight / height ** 2
print(bmi)
if bmi < 18.5:
    print('Underweight')
elif bmi > 18.5 and bmi < 25:
    print('Normal weight')
elif bmi > 25 and bmi < 30:
    print('Overweight') 
else:
    print('Obese')
```

### Task #3: Use while loop to get Sum of squared integers (1 point)

Your code should print 8 as the last integer to be squared and 204 as the sum of the squared integers. 

```python
# Write your code here - insert more code cells as needed
sum = 0
i = 1
while True:
    sum += i ** 2
    if sum > 200:
        break
    i += 1
print(i)
print(sum)
```

### Task #4: Using while loop write program to Guess the secret word (2 points)
Write code that keeps prompting the user to guess a word. The user is allowed up to 5 guesses until the program stops prompting the user and the word is revealed. A while loop must be utilized to complete this task.

Additional Criteria:

- The word may be hardcoded in a variable
- The maximum number of guess attempts may be hardcoded in a variable
- You must use the input function for the user to input their guess
- If the user guesses the word correctly, the program will print a message that the user is correct and what the word is. The program will no longer prompt the user to guess the word
- If the user guesses incorrectly (for attempts 1, 2, 3, & 4), the program will print a message that the guess is incorrect and prompt the user for another guess
- If the user's 5th attempt is incorrect, the program will print a message that the guess is incorrect, that the user has no more guess attempts and reveal what the word is. The program will also no longer prompt the user to guess the word

```python
# Write your code here - insert more code cells as needed
WORD = 'python'
MAX = 5
i = 1
while i <= MAX:
    guess = input('Guess the word: ')
    if guess == WORD:
        break
    print('Your guess is incorrect.')
    i +=1
if i > MAX:
    print(f'You are out of guesses. The word is {WORD}.')
else:
    print(f'Correct. The word is {WORD}.')
```

### Task #6: Use while loop to get Multiples of 4 and total (2 points)
Write code that prints numbers that are multiples of 4, from 40 to 4 (inclusive) in descending order. A while loop must be utilized to complete this task.

Your code should print the following:

40
36
32
28
24
20
16
12
8
4

##### In addition, your code should also print the sum of all numbers above EXCLUDING numbers that are multiples of 10. Meaning that all numbers listed above would be included except for 20 & 40.

Your code should print 160 as the answer.

```python
# Write your code here - insert more code cells as needed
i = 40
sum  = 0
STEP = LAST = 4
while i >= LAST:
    print(F'{i} ', end='')
    if i % 10 != 0:
        sum += i
    i -= STEP
print(f'\nSum: {sum}')
```

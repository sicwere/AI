```python
import warnings
warnings.filterwarnings("ignore")
from pathlib import Path
import whisper
from retinaface import RetinaFace
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

def face_recognition():
    vid = cv2.VideoCapture(vid_path)
    while True:
        ret, frame = vid.read()
        if not ret:
            break
        faces = RetinaFace.detect_faces(frame)
        if len(faces) > 0:
            for face in faces.items():
                x1, y1, x2, y2 = face[1]['facial_area'] 
                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2) 
        #cv2.imshow('Video', frame)
        plt.imshow(frame)
        plt.show()
    vid.release()
    cv2.destroyAllWindows()

def object_recognition():
    model = YOLO("yolo26n.pt")
    vid = cv2.VideoCapture(vid_path)
    while True:
        ret, frame = vid.read()
        if not ret:
            break
        results = model(frame)
        for result in results:
            for box in result.boxes:
                # Get the class ID of the detected object
                class_id = int(box.cls[0])
                # Get the class name from the model's metadata
                obj = model.names[class_id]
                print(obj)

def speech_recognition():
    model = whisper.load_model('base')
    result = model.transcribe(vid_path)
    print(result['text'])

vid_path = input('Enter file path of video:')
file_path = Path(vid_path)
if not file_path.is_file():
    print('File doesn\'t exist.')
else:
    choice = ''
    while choice != 'Q':
        choice = input("Press 'F' for face detection, 'O' for object detection, 'S' for speech detection, or 'Q' to quit:").upper()
        match choice:
            case 'F':
                face_recognition()
            case 'O':
                object_recognition()
            case 'S':
                speech_recognition()
```
